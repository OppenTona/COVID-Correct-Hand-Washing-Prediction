## 1. Import required Libraries

```python
import os
import cv2
import shutil
```

## 2. Image Resizing

```python
# Dataset đã chứa toàn bộ ảnh StageN sau khi copy
dataset_path = 'C:/Users/Admin/OneDrive - University of Huddersfield/Data-driven-AI/Assignment 2/dataset'

processed_image_count = 0
count_invalid_size = 0
invalid_size_images = []

output_file = 'image_labels.txt'
invalid_size_file = 'invalid_size_images.txt'

labels = []

print("===== CHECKING IMAGE SIZE + ASSIGNING LABELS =====\n")

# ==============================================
# 1. XÁC ĐỊNH LABEL MỖI ẢNH DỰA TRÊN TÊN THƯ MỤC StageN
# ==============================================

for folder_name in sorted(os.listdir(dataset_path)):

    folder_path = os.path.join(dataset_path, folder_name)

    if not os.path.isdir(folder_path):
        continue

    # Folder phải là dạng StageN
    if folder_name.lower().startswith("stage"):
        try:
            stage_number = int(folder_name[5:])
        except:
            print(f"⚠️ Folder không hợp lệ: {folder_name}")
            continue

        print(f"\n🔍 Processing folder: {folder_name} (label = {stage_number -1})")

        # ==============================================
        # 2. DUYỆT ẢNH TRONG TỪNG STAGE
        # ==============================================

        for filename in os.listdir(folder_path):

            if not filename.lower().endswith(('.jpg', '.jpeg', '.png')):
                print(f"⚠️ Skipping non-image: {filename}")
                continue

            img_path = os.path.join(folder_path, filename)
            img = cv2.imread(img_path)

            if img is None:
                print(f"❌ Cannot read (corrupted): {img_path}")
                continue

            h, w = img.shape[:2]

            # Kiểm tra kích thước
            if (h != 150) or (w != 150):
                invalid_size_images.append(f"{img_path} | SIZE = {w}x{h}")
                count_invalid_size += 1
                print(f"⚠️ Invalid size: {img_path} → {w}x{h} (required 150x150)")
                continue  # Không gán label cho ảnh sai kích thước

            # Ảnh hợp lệ → gán label
            labels.append((filename, stage_number - 1))
            processed_image_count += 1

    else:
        print(f"⚠️ Ignoring non-Stage folder: {folder_name}")

# ==============================================
# 3. SAVE LABEL FILES
# ==============================================

with open(output_file, "w", encoding="utf-8") as f:
    for filename, label in labels:
        f.write(f"{filename}: {label}\n")

print(f"\n🎯 Saved {len(labels)} labeled images to {output_file}")

# Lưu ảnh sai kích thước
with open(invalid_size_file, "w", encoding="utf-8") as f:
    for item in invalid_size_images:
        f.write(item + "\n")

print(f"⚠️ {count_invalid_size} images have invalid size. Saved list to {invalid_size_file}")

# ==============================================
# 4. REPORT
# ==============================================

print("\n===== SUMMARY REPORT =====")
print("Valid processed images:", processed_image_count)
print("Invalid-size images:", count_invalid_size)
print("Total images scanned:", processed_image_count + count_invalid_size)

if count_invalid_size == 0:
    print("\n🎉 All images are valid 150x150!")
```

```python
# ==============================================
# 1. PATH SETUP
# ==============================================

repo_path = 'C:/Users/Admin/OneDrive - University of Huddersfield/Data-driven-AI/Assignment 2/CHS2406_Coursework2_Data_Repository'
dataset_path = 'C:/Users/Admin/OneDrive - University of Huddersfield/Data-driven-AI/Assignment 2/dataset'

processed_image_count = 0
count_label_loading_error = 0
count_invalid_size = 0  # đếm ảnh sai kích thước
invalid_size_images = []  # lưu tên ảnh sai kích thước

os.makedirs(dataset_path, exist_ok=True)

output_file = 'image_labels.txt'
invalid_size_file = 'invalid_size_images.txt'

labels = []
corrupted_images = []
unlabeled_images = []

print("===== COPYING IMAGES & ASSIGNING LABELS =====\n")

# ==============================================
# 2. MAIN LOOP: COPY + LABEL + CHECK SIZE
# ==============================================

label_map = {}  # StageN → N

for folder_name in sorted(os.listdir(repo_path)):
    source_folder = os.path.join(repo_path, folder_name)

    if not os.path.isdir(source_folder):
        continue  # skip file tại root

    if folder_name.lower().startswith("stage"):
        try:
            stage_number = int(folder_name[5:])
            label_map[folder_name] = stage_number
        except:
            print(f"⚠️ Warning: Folder {folder_name} is not valid StageN.")
            continue

        current_label_int = label_map[folder_name]

        for filename in os.listdir(source_folder):
            source_file = os.path.join(source_folder, filename)

            if not filename.lower().endswith(('.jpg', '.jpeg', '.png')):
                print(f"⚠️ Skipping non-image file: {source_file}")
                count_label_loading_error += 1
                continue

            img = cv2.imread(source_file)

            if img is None:
                corrupted_images.append(os.path.join(folder_name, filename))
                print(f"❌ Cannot read (corrupted): {source_file}")
                count_label_loading_error += 1
                continue

            # Kiểm tra kích thước ảnh
            if img.shape[0] != 150 or img.shape[1] != 150:
                invalid_size_images.append(os.path.join(folder_name, filename))
                count_invalid_size += 1
                print(f"⚠️ Invalid size (not 150x150): {source_file}")
                count_label_loading_error += 1  # bạn có thể tách riêng nếu muốn
                continue  # bỏ qua ảnh không đúng kích thước

            # Copy ảnh hợp lệ vào dataset
            dst = os.path.join(dataset_path, filename)
            shutil.copy2(source_file, dst)
            processed_image_count += 1

            # Gán label ngay lập tức
            labels.append((filename, current_label_int-1))

    else:
        # Folder không phải StageN → ảnh không gán nhãn
        for filename in os.listdir(source_folder):
            if filename.lower().endswith(('.jpg', '.jpeg', '.png')):
                unlabeled_images.append(os.path.join(folder_name, filename))


# ==============================================
# 3. SAVE LABEL FILE
# ==============================================

with open(output_file, "w", encoding="utf-8") as f:
    for filename, label in labels:
        f.write(f"{filename}: {label}\n")

print(f"\n🎯 Saved {len(labels)} labeled images to {output_file}")

# Lưu danh sách ảnh sai kích thước
with open(invalid_size_file, "w", encoding="utf-8") as f:
    for path in invalid_size_images:
        f.write(f"{path}\n")

print(f"⚠️ {count_invalid_size} images have invalid size. Saved list to {invalid_size_file}")

# ==============================================
# 4. REPORT
# ==============================================

print("\nTotal successfully processed images: ", processed_image_count)
print("Number of unreadable images or errors: ", count_label_loading_error)
print("Total images scanned: ", processed_image_count + count_label_loading_error)

if corrupted_images:
    print("\n❌ Corrupted images:")
    for x in corrupted_images:
        print(" -", x)

if unlabeled_images:
    print("\n⚠️ Images in non-Stage folders (not labeled):")
    for x in unlabeled_images:
        print(" -", x)

if not corrupted_images and not unlabeled_images and count_invalid_size == 0:
    print("\n🎉 All images were valid and assigned labels correctly!")
```
