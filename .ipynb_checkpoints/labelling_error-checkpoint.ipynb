## 2. Image Resizing

```python
import os
import cv2

dataset_path = 'C:/Users/Admin/OneDrive - University of Huddersfield/Data-driven-AI/Assignment 2/dataset'

count_invalid_size = 0
invalid_size_images = []

invalid_size_file = 'invalid_size_images.txt'

print("===== CHECKING IMAGE SIZE IN DATASET =====\n")

# Duyệt toàn bộ file trong dataset (chỉ có ảnh)
for filename in sorted(os.listdir(dataset_path)):

    # Chỉ xử lý file ảnh
    if not filename.lower().endswith(('.jpg', '.jpeg', '.png')):
        print(f"⚠️ Skipping non-image: {filename}")
        continue

    img_path = os.path.join(dataset_path, filename)
    img = cv2.imread(img_path)

    # Kiểm tra đọc ảnh thất bại
    if img is None:
        print(f"❌ Cannot read (corrupted): {img_path}")
        continue

    h, w = img.shape[:2]

    # Kiểm tra kích thước
    if (h != 150) or (w != 150):
        count_invalid_size += 1
        invalid_size_images.append(f"{filename}  |  SIZE = {w}x{h}")
        print(f"⚠️ Invalid size: {filename} → {w}x{h} (required 150x150)")

print("\n===== SUMMARY REPORT =====")
print("Total images scanned:", len([f for f in os.listdir(dataset_path) if f.lower().endswith(('.jpg','.jpeg','.png'))]))
print("Invalid-size images:", count_invalid_size)

# Lưu file danh sách ảnh sai kích thước
with open(invalid_size_file, "w", encoding="utf-8") as f:
    for item in invalid_size_images:
        f.write(item + "\n")

print(f"\n⚠️ Saved list of invalid-size images to {invalid_size_file}")

if count_invalid_size == 0:
    print("\n🎉 All images in dataset are 150x150!")
```

```python
dataset_path = 'C:/Users/Admin/OneDrive - University of Huddersfield/Data-driven-AI/Assignment 2/dataset'

invalid_size_file = 'invalid_size_images.txt'
label_file = 'image_labels.txt'

print("===== CHECKING IMAGE SIZE BY LABEL =====\n")

# ===========================================
# 1. LOAD LABELS FROM image_labels.txt
# ===========================================

label_map = {}   # filename → label

with open(label_file, "r", encoding="utf-8") as f:
    for line in f:
        if ":" in line:
            fname, lbl = line.strip().split(":")
            label_map[fname.strip()] = int(lbl.strip())


# ===========================================
# 2. BIẾN ĐẾM THEO TỪNG LABEL (Stage)
# ===========================================

invalid_count_per_label = {}    # label → số ảnh sai size
invalid_size_images = []        # lưu toàn bộ ảnh sai size

# Tạo key sẵn cho các label xuất hiện trong file
for lbl in set(label_map.values()):
    invalid_count_per_label[lbl] = 0


# ===========================================
# 3. KIỂM TRA KÍCH THƯỚC ẢNH
# ===========================================

for filename in sorted(os.listdir(dataset_path)):

    # Bỏ qua file không phải ảnh
    if not filename.lower().endswith(('.jpg', '.jpeg', '.png')):
        continue

    img_path = os.path.join(dataset_path, filename)
    img = cv2.imread(img_path)

    if img is None:
        print(f"❌ Cannot read (corrupted): {img_path}")
        continue

    h, w = img.shape[:2]

    # Lấy label của ảnh từ file label
    if filename not in label_map:
        print(f"⚠️ WARNING: {filename} not found in label file.")
        continue

    label = label_map[filename]

    # Ảnh sai kích thước
    if h != 150 or w != 150:
        invalid_count_per_label[label] += 1
        invalid_size_images.append(f"{filename} | LABEL={label} | SIZE={w}x{h}")
        print(f"⚠️ Invalid size: {filename} (label={label}) → {w}x{h}")


# ===========================================
# 4. SUMMARY REPORT
# ===========================================

print("\n===== SUMMARY REPORT =====")
print("Invalid-size images (total):", sum(invalid_count_per_label.values()))

print("\nInvalid-count per label (Stage):")
for lbl in sorted(invalid_count_per_label.keys()):
    print(f" - Label {lbl}: {invalid_count_per_label[lbl]} images")

# Lưu danh sách chi tiết
with open(invalid_size_file, "w", encoding="utf-8") as f:
    for item in invalid_size_images:
        f.write(item + "\n")

print(f"\n⚠️ Saved list of invalid-size images to {invalid_size_file}")

if sum(invalid_count_per_label.values()) == 0:
    print("\n🎉 All images are valid 150x150!")
```

```python
import os
import cv2

dataset_path = 'C:/Users/Admin/OneDrive - University of Huddersfield/Data-driven-AI/Assignment 2/dataset'

count_invalid_size = 0

print("===== CHECKING AND FIXING IMAGE SIZE IN DATASET =====\n")

# Duyệt toàn bộ file trong dataset (chỉ có ảnh)
for filename in sorted(os.listdir(dataset_path)):

    # Chỉ xử lý file ảnh
    if not filename.lower().endswith(('.jpg', '.jpeg', '.png')):
        print(f"⚠️ Skipping non-image: {filename}")
        continue

    img_path = os.path.join(dataset_path, filename)
    img = cv2.imread(img_path)

    if img is None:
        print(f"❌ Cannot read (corrupted): {img_path}")
        continue

    h, w = img.shape[:2]

    # Kiểm tra kích thước
    if (h != 150) or (w != 150):
        count_invalid_size += 1
        print(f"⚠️ Fixing size: {filename} → {w}x{h} (resizing to 150x150)")

        # Resize về 150x150
        resized = cv2.resize(img, (150, 150), interpolation=cv2.INTER_AREA)

        # Ghi đè lên file cũ
        cv2.imwrite(img_path, resized)

print("\n===== SUMMARY REPORT =====")
print("Total images scanned:", len([f for f in os.listdir(dataset_path) if f.lower().endswith(('.jpg','.jpeg','.png'))]))
print("Images resized to correct size:", count_invalid_size)

if count_invalid_size == 0:
    print("\n🎉 All images in dataset were already 150x150!")
else:
    print("\n✨ All invalid images have been resized successfully!")
```
