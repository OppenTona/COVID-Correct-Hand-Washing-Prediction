## 1. Import required Libraries

```python
import os
from PIL import Image
from collections import Counter
import cv2
import numpy as np
import pandas as pd
```

## 2. Cleanlab

```python
labels_file = "image_labels.txt"
dataset_path = "dataset"
required_size = (150, 150)

# ---- Load label file ----
labels = []
with open(labels_file, "r", encoding="utf-8") as f:
    for line in f:
        parts = line.strip().split(": ")
        if len(parts) == 2:
            labels.append((parts[0], int(parts[1])))

label_map = {fname: lbl for fname, lbl in labels}

# ---- Load all images into memory ----
X_images = []
y_labels = []
filenames = []

for filename in os.listdir(dataset_path):
    if not filename.lower().endswith((".jpg", ".jpeg", ".png")):
        continue

    label = label_map.get(filename)
    if label is None:
        continue

    img_path = os.path.join(dataset_path, filename)
    try:
        img = Image.open(img_path).convert("RGB")
        img = img.resize(required_size)
        X_images.append(np.array(img))
        y_labels.append(label)
        filenames.append(filename)
    except Exception as e:
        print(f"Lỗi load ảnh {filename}: {e}")

X_images = np.array(X_images, dtype=np.float32)
y_labels = np.array(y_labels)

print("✔ Loaded images:", X_images.shape)
print("✔ Loaded labels:", y_labels.shape)
```

```python
import tensorflow as tf

print("✨ Extracting MobileNetV2 embeddings...")

base_model = tf.keras.applications.MobileNetV2(
    weights="imagenet",
    include_top=False,
    pooling="avg",
    input_shape=(150,150,3)
)

X_pre = tf.keras.applications.mobilenet_v2.preprocess_input(X_images)
X_features = base_model.predict(X_pre, verbose=1)

print("✔ Feature shape:", X_features.shape)
```

```python
from xgboost import XGBClassifier

print("🚀 Training XGBoost classifier...")

clf = XGBClassifier(
    n_estimators=500,
    learning_rate=0.05,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="multi:softprob",
    eval_metric="mlogloss"
)

clf.fit(X_features, y_labels)

pred_probs = clf.predict_proba(X_features)
print("✔ Predicted prob shape:", pred_probs.shape)
```

```python
from cleanlab.filter import find_label_issues

print("🔍 Running Cleanlab...")

issues = find_label_issues(
    labels=y_labels,
    pred_probs=pred_probs,
    return_indices_ranked_by="self_confidence"
)

print(f"⚠️ Tổng số ảnh nghi sai nhãn: {len(issues)}")
```
